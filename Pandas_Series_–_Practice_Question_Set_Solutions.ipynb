```python
#28 Without running the code, trace the complete transformation and state the index and values of s2. Then state whether s and s2 represent the same labelled data.
s = pd.Series([90, 75, 82],index=['Ravi', 'Aman', 'Neha'])

d = s.to_dict()
s2 = pd.Series(d)

print("Original Series:")
print(s)

print("\nDictionary:")
print(d)

print("\nNew Series:")
print(s2)

print("\nAre they equal?")
print(s.equals(s2))
```

```python
#29 Write a short Pandas program that creates a labelled Series for four products and sales, displays its values and index, selects one label, filters sales above 500, checks whether a label exists, and converts the Series to a dictionary.
sales = pd.Series([450, 700, 850, 300],index=['Pen', 'Notebook', 'Bag', 'Pencil'])
print("Values:")
print(sales.values)
print("\nIndex:")
print(sales.index)
print("\nNotebook Sales:")
print(sales['Notebook'])
print("\nSales above 500:")
print(sales[sales > 500])
print("\nIs Bag present?")
print('Bag' in sales.index)
sales_dict = sales.to_dict()
print("\nDictionary:")
print(sales_dict)
```

```python
#30 Analyse the following code and answer all parts: (a) What is the index order of s? (b) Which label contains NaN? (c) What will the Boolean filter s[s > 20] return? (d) What will pd.notnull(s) indicate for each label? (e) What are the Series name and index name after the final two statements?
data = {'A': 15,'B': 35,'C': 25}
s = pd.Series(data,index=['C', 'A', 'D', 'B'])
s.name = 'Observations'
s.index.name = 'Category'
print("Index order:")
print(s.index)
print("\nSeries:")
print(s)
print("\nValues greater than 20:")
print(s[s > 20])
print("\nNon-null values:")
print(pd.notnull(s))
print("\nSeries name:")
print(s.name)
print("\nIndex name:")
print(s.index.name)
```
